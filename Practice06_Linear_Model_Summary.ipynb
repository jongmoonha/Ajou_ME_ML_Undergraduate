# Practice 06 — Linear Model Summary

–04에서 다룬 4가지 선형 모델의 **학습 코드**를 한 곳에 정리합니다.

| # | Model | Activation | $\hat{\mathbf{y}} =$ | Loss | $\mathbf{E}$ (잔차) | $\nabla J$ | Update |
|---|---|---|---|---|---|---|---|
| 1 | Linear Regression | identity | $\mathbf{Xw}$ | MSE $\|\mathbf{E}\|^2$ | $\mathbf{y}-\hat{\mathbf{y}}$ | $-2\mathbf{X}^T\mathbf{E}$ | $\mathbf{w} \leftarrow \mathbf{w} + 2\rho\,\mathbf{X}^T\mathbf{E}$ |
| 2 | Perceptron | unit step | $\text{step}(\mathbf{Xw})$ | miscount | $\mathbf{y}-\hat{\mathbf{y}}$ | $-\mathbf{X}^T\mathbf{E}$ | $\mathbf{w} \leftarrow \mathbf{w} + \rho\,\mathbf{X}^T\mathbf{E}$ |
| 3 | Logistic (binary) | sigmoid | $\sigma(\mathbf{Xw})$ | BCE | $\mathbf{y}-\hat{\mathbf{y}}$ | $-\mathbf{X}^T\mathbf{E}$ | $\mathbf{w} \leftarrow \mathbf{w} + \rho\,\mathbf{X}^T\mathbf{E}$ |
| 4 | Logistic (multi) | softmax | $\text{softmax}(\mathbf{XW}) = \mathbf{O}$ | CE | $\mathbf{Y}-\mathbf{O}$ | $-\mathbf{X}^T\mathbf{E}$ | $\mathbf{W} \leftarrow \mathbf{W} + \rho\,\mathbf{X}^T\mathbf{E}$ |

→ 모든 모델이 공통적으로 $\nabla J = -\mathbf{X}^T \mathbf{E}$ 형태 (잔차 $\mathbf{E}$ 는 모델별 "target − pred").

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets

plt.rcParams['figure.figsize'] = (10, 4)
plt.rcParams['axes.unicode_minus'] = False
```

---
# 1. Linear Regression

$$\hat{\mathbf{y}} = \mathbf{Xw}, \qquad J = \|\mathbf{y} - \hat{\mathbf{y}}\|^2, \qquad \frac{\partial J}{\partial \mathbf{w}} = -2\mathbf{X}^T(\mathbf{y} - \hat{\mathbf{y}})$$

### 데이터 로드

```python
# sklearn 당뇨병 데이터셋 (1개 feature만 사용)
diabetes = datasets.load_diabetes()
x = diabetes.data[:, 2:3]          # BMI feature, shape (442, 1)
y = diabetes.target.reshape(-1, 1)  # shape (442, 1)

# Augmented matrix: [1, x]
X = np.hstack([np.ones((len(x), 1)), x])

print(f"X shape: {X.shape}, y shape: {y.shape}")
```

```python
# 데이터 미리보기
pd.DataFrame({'BMI': x.flatten(), 'target': y.flatten()}).head(10)
```

### 학습 파라미터

```python
w = np.zeros((2, 1))
rho = 0.001
n_iter = 3000
```

### 학습

```python
loss_hist = []

for i in range(n_iter):
    y_hat = X @ w
    E = y - y_hat
    dJ = -2 * X.T @ E
    w  = w - rho * dJ
    loss_hist.append(float(np.sum(E**2)))

print(f"w0 = {w[0,0]:.2f}, w1 = {w[1,0]:.2f}")
```

### 평가 + 시각화

```python
# R² (결정 계수)
y_pred = X @ w
ss_res = np.sum((y - y_pred)**2)
ss_tot = np.sum((y - y.mean())**2)
r2 = 1 - ss_res / ss_tot
print(f"R² = {r2:.4f}")

fig, axes = plt.subplots(1, 2)
axes[0].scatter(x, y, c='r', s=10, alpha=0.5)
axes[0].plot(x, X @ w, 'b-', lw=2)
axes[0].set_title(f'Linear Regression (R$^2$={r2:.3f})'); axes[0].set_xlabel('BMI'); axes[0].set_ylabel('target'); axes[0].grid(alpha=0.3)
axes[1].plot(loss_hist, 'b-', lw=1)
axes[1].set_title('MSE Loss'); axes[1].set_xlabel('iter'); axes[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

---
# 2. Perceptron

$$\hat{\mathbf{y}} = \text{step}(\mathbf{Xw}), \qquad \frac{\partial J}{\partial \mathbf{w}} = -\mathbf{X}^T(\mathbf{y}-\hat{\mathbf{y}}), \qquad \mathbf{w} \leftarrow \mathbf{w} + \rho\, \mathbf{X}^T(\mathbf{y}-\hat{\mathbf{y}})$$

### 데이터 로드

```python
# sklearn Iris 데이터셋 (2 classes, 2 features)
iris = datasets.load_iris()
mask = iris.target < 2                       # class 0, 1만 사용
X_raw = iris.data[mask, :2]                  # sepal length, sepal width
y_cls = iris.target[mask]                    # 0 or 1

# Augmented matrix: [1, x1, x2]
X_cls = np.column_stack([np.ones(len(y_cls)), X_raw])

print(f"X shape: {X_cls.shape}, classes: {np.unique(y_cls)}")
```

```python
# 데이터 미리보기
pd.DataFrame({
    'sepal length': X_raw[:, 0], 'sepal width': X_raw[:, 1],
    'y': y_cls, 'class': [iris.target_names[v] for v in y_cls]
}).head(10)
```

### 학습 파라미터

```python
step = lambda z: (z >= 0).astype(int)

w = np.zeros(3)
rho = 0.1
n_epochs = 100
```

### 학습

```python
acc_hist = []

for ep in range(n_epochs):
    y_hat = step(X_cls @ w)
    E = y_cls - y_hat
    acc_hist.append(np.mean(y_hat == y_cls))
    dJ = -X_cls.T @ E
    w = w - rho * dJ
    if acc_hist[-1] == 1.0:
        break

print(f"Epochs: {len(acc_hist)}, Accuracy: {acc_hist[-1]:.1%}")
print(f"w = {np.round(w, 3)}")
```

### 평가 + 시각화

```python
fig, axes = plt.subplots(1, 2)
pos, neg = y_cls==1, y_cls==0
axes[0].scatter(X_raw[pos,0], X_raw[pos,1], c='b', s=30, label='y=1', alpha=0.7)
axes[0].scatter(X_raw[neg,0], X_raw[neg,1], c='r', s=30, label='y=0', alpha=0.7)
x1_line = np.linspace(X_raw[:,0].min(), X_raw[:,0].max(), 100)
x2_line = -(w[0] + w[1]*x1_line) / w[2]
axes[0].plot(x1_line, x2_line, 'k--', lw=2)
axes[0].set_title(f'Perceptron (Acc={acc_hist[-1]:.1%})'); axes[0].set_xlabel('sepal length'); axes[0].set_ylabel('sepal width')
axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(range(1, len(acc_hist)+1), [a*100 for a in acc_hist], 'g-', lw=1.5)
axes[1].set_title(f'Accuracy (final: {acc_hist[-1]:.1%})')
axes[1].set_xlabel('epoch'); axes[1].set_ylabel('Accuracy (%)')
axes[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

# 3. Logistic Regression (Binary)

$$\hat{\mathbf{y}} = \sigma(\mathbf{Xw}) = \frac{1}{1+e^{-\mathbf{Xw}}}, \qquad J = -\sum_n \bigl[ y_n \log \hat{y}_n + (1-y_n) \log(1-\hat{y}_n) \bigr]$$

$$\frac{\partial J}{\partial \mathbf{w}} = -\mathbf{X}^T(\mathbf{y} - \hat{\mathbf{y}}), \qquad \mathbf{w} \leftarrow \mathbf{w} + \rho\, \mathbf{X}^T(\mathbf{y}-\hat{\mathbf{y}})$$

### 데이터 로드

```python
# Iris 동일 데이터 사용 (위 Section 2와 같음)
print(f"X shape: {X_cls.shape}, classes: {np.unique(y_cls)}")
```

### 학습 파라미터

```python
sigmoid = lambda z: 1 / (1 + np.exp(-np.clip(z, -500, 500)))

w = np.zeros(3)
rho = 0.01
n_epochs = 200
```

### 학습

```python
acc_hist = []

for ep in range(n_epochs):
    y_hat = sigmoid(X_cls @ w)
    E = y_cls - y_hat
    acc_hist.append(np.mean((y_hat >= 0.5).astype(int) == y_cls))
    loss = -np.sum(y_cls * np.log(y_hat+1e-15) + (1-y_cls) * np.log(1-y_hat+1e-15))
    dJ = -X_cls.T @ E
    w = w - rho * dJ

print(f"w = {np.round(w, 3)}")
print(f"Accuracy: {acc_hist[-1]:.1%}")
```

### 평가 + 시각화

```python
fig, axes = plt.subplots(1, 2)
axes[0].scatter(X_raw[pos,0], X_raw[pos,1], c='b', s=30, label='y=1', alpha=0.7)
axes[0].scatter(X_raw[neg,0], X_raw[neg,1], c='r', s=30, label='y=0', alpha=0.7)
x1_line = np.linspace(X_raw[:,0].min(), X_raw[:,0].max(), 100)
x2_line = -(w[0] + w[1]*x1_line) / w[2]
axes[0].plot(x1_line, x2_line, 'k--', lw=2)
axes[0].set_title(f'Logistic Regression (Acc={acc_hist[-1]:.1%})'); axes[0].set_xlabel('sepal length'); axes[0].set_ylabel('sepal width')
axes[0].legend(); axes[0].grid(alpha=0.3)

axes[1].plot(range(1, n_epochs+1), [a*100 for a in acc_hist], 'g-', lw=1.5)
axes[1].set_title(f'Accuracy (final: {acc_hist[-1]:.1%})')
axes[1].set_xlabel('epoch'); axes[1].set_ylabel('Accuracy (%)')
axes[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

# 4. Logistic Regression (Multi-class)

$$\mathbf{O} = \text{softmax}(\mathbf{XW}), \qquad J = -\sum_n \log o_{y_n}^{(n)}$$

$$\frac{\partial J}{\partial \mathbf{W}} = -\mathbf{X}^T(\mathbf{Y} - \mathbf{O}), \qquad \mathbf{W} \leftarrow \mathbf{W} + \rho\, \mathbf{X}^T(\mathbf{Y}-\mathbf{O})$$

$\mathbf{W}$: shape $(d+1) \times C$ (행=입력, 열=클래스), $\mathbf{Y}$: one-hot target matrix, $\mathbf{O}$: softmax 예측 확률.

### 데이터 로드

```python
# Iris 3-class, 4 features
iris = datasets.load_iris()
X_raw_mc = iris.data           # (150, 4)
y_mc = iris.target             # (150,), 0/1/2
N, d = X_raw_mc.shape
C = 3

X_mc = np.column_stack([np.ones(N), X_raw_mc])   # (150, 5)

# One-hot encoding
y_onehot = np.zeros((N, C))
y_onehot[np.arange(N), y_mc] = 1

print(f"X shape: {X_mc.shape}, classes: {np.unique(y_mc)}, N: {N}")
```

```python
# 데이터 미리보기
pd.DataFrame(
    X_raw_mc, columns=iris.feature_names
).assign(y=y_mc, **{'class': [iris.target_names[v] for v in y_mc]}).head(10)
```

### 학습 파라미터

```python
def softmax(z):
    e = np.exp(z - z.max(axis=-1, keepdims=True))
    return e / e.sum(axis=-1, keepdims=True)

W = np.zeros((d+1, C))   # (5, 3): 행=입력, 열=클래스
rho = 0.001
n_epochs = 300
```

### 학습

```python
loss_hist, acc_hist = [], []

for ep in range(n_epochs):
    O = softmax(X_mc @ W)                                    # (N, C)
    E = y_onehot - O                                         # 잔차 (N, C)
    loss = -np.sum(np.log(O[np.arange(N), y_mc] + 1e-15))
    acc  = np.mean(np.argmax(O, axis=1) == y_mc)
    loss_hist.append(loss)
    acc_hist.append(acc)
    dW = -X_mc.T @ E                                         # (d+1, C)
    W  = W - rho * dW

print(f"Loss: {loss_hist[0]:.2f} → {loss_hist[-1]:.2f}")
print(f"Accuracy: {acc_hist[0]:.1%} → {acc_hist[-1]:.1%}")
```

### 평가 + 시각화

```python
fig, axes = plt.subplots(1, 2)

axes[0].plot(range(1, n_epochs+1), loss_hist, "b-", lw=1.5)
axes[0].set_title("CE Loss"); axes[0].set_xlabel("epoch"); axes[0].grid(alpha=0.3)

axes[1].plot(range(1, n_epochs+1), [a*100 for a in acc_hist], "g-", lw=1.5)
axes[1].set_title("Accuracy (%)"); axes[1].set_xlabel("epoch"); axes[1].grid(alpha=0.3)

plt.tight_layout(); plt.show()
```

# 5. 비교 요약

| | Linear Regression | Perceptron | Logistic (Binary) | Logistic (Multi) |
|---|---|---|---|---|
| **Activation** | identity | step | $\sigma(z)=\frac{1}{1+e^{-z}}$ | softmax |
| **$\hat{\mathbf{y}}$** | $\mathbf{Xw}$ | $\text{step}(\mathbf{Xw})$ | $\sigma(\mathbf{Xw})$ | $\text{softmax}(\mathbf{XW}) = \mathbf{O}$ |
| **Loss** | $\|\mathbf{y}-\hat{\mathbf{y}}\|^2$ | miscount | $-\sum [y\log\hat{y} + (1-y)\log(1-\hat{y})]$ | $-\sum \log o_{y_n}$ |
| **잔차 $\mathbf{E}$** | $\mathbf{y}-\hat{\mathbf{y}}$ | $\mathbf{y}-\hat{\mathbf{y}}$ | $\mathbf{y}-\hat{\mathbf{y}}$ | $\mathbf{Y}-\mathbf{O}$ |
| **Gradient** | $-2\mathbf{X}^T\mathbf{E}$ | $-\mathbf{X}^T\mathbf{E}$ | $-\mathbf{X}^T\mathbf{E}$ | $-\mathbf{X}^T\mathbf{E}$ |
| **Update** | $\mathbf{w} + 2\rho\,\mathbf{X}^T\mathbf{E}$ | $\mathbf{w} + \rho\,\mathbf{X}^T\mathbf{E}$ | $\mathbf{w} + \rho\,\mathbf{X}^T\mathbf{E}$ | $\mathbf{W} + \rho\,\mathbf{X}^T\mathbf{E}$ |
| **Output** | $\mathbb{R}$ | $\{0,1\}$ | $(0,1)$ | $(0,1)^C$ |
| **Dataset** | Diabetes | Iris (2-class) | Iris (2-class) | Iris (3-class, 4 features) |

→ 공통 형태: $\nabla J = -\mathbf{X}^T \mathbf{E}$, &nbsp; Update: $\mathbf{w} \leftarrow \mathbf{w} + \rho\,\mathbf{X}^T \mathbf{E}$ (Linear만 $2\rho$)

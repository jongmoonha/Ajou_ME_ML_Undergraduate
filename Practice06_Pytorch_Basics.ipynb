# Practice 06 — Pytorch 기초

Pytorch의 **Tensor**와 **Autograd**를 단계별로 익힙니다.

| 핵심 개념 | 설명 |
|---|---|
| `torch.tensor(x, requires_grad=True)` | 미분 추적 활성화 |
| `y.backward()` | $dy/dw$ 자동 계산 |
| `w.grad` | 계산된 그래디언트 값 조회 |

````python
import numpy as np
import torch
import matplotlib.pyplot as plt
````

---
# 1. Tensor 기초

PyTorch의 핵심 자료형인 **Tensor**는 NumPy 배열과 유사하지만,  
`requires_grad=True`를 설정하면 **자동 미분(autograd)**이 가능합니다.

> 여기서는 autograd 에 필요한 최소한의 Tensor 생성법만 확인합니다 (shape 변환·결합·변환 등 다양한 조작법은 별도로 다룹니다).

````python
# tensor 생성: requires_grad=True → 이 텐서에 대한 미분을 추적
t = torch.tensor(2.0, requires_grad=True)
print(t)
print(f'값: {t.item()},  requires_grad: {t.requires_grad}')

# NumPy 배열 → Tensor
arr = np.array([1.0, 2.0, 3.0])
t_from_np = torch.tensor(arr, dtype=torch.float32)
print(f'\nNumPy → Tensor: {t_from_np}')
````

---
# 2. Autograd — 자동 미분

Forward 연산을 정의하면, `backward()`로 그래디언트를 자동 계산합니다.

### 예제 1: $y = w^2 + 2w + 1$

$$y = w^2 + 2w + 1 \qquad \Rightarrow \qquad \frac{dy}{dw} = 2w + 2$$

````python
w = 2 # Try different values
w_tensor = torch.tensor(w, dtype=float, requires_grad=True)
y_tensor = w_tensor**2 + 2*w_tensor + 1

# calculate the gradient
y_tensor.backward()

print('w: ', w_tensor)
print('y: ', y_tensor)
print('grad: ', w_tensor.grad)   # 2*2+2 = 6
````

### 예제 2: $\hat{y} = 3w + 1$

$$\hat{y} = 3w + 1 \qquad \Rightarrow \qquad \frac{d\hat{y}}{dw} = 2$$

````python
w = 2 # Try different values
w_tensor = torch.tensor(w, dtype=float, requires_grad=True)
y_tensor = 3*w_tensor + 1

# calculate the gradient
y_tensor.backward()

print('w: ', w_tensor)
print('y: ', y_tensor)
print('grad: ', w_tensor.grad)   # 2
````

### 예제 3: $\hat{y} = 3w + 1, \; f = \hat{y}^2$

$$\hat{y} = xw + 1, \qquad f = \hat{y}^2$$

$$\frac{d\hat{y}}{dw} = x, \qquad \frac{df}{d\hat{y}} = 2\hat{y}$$

$$\frac{df}{dw} = x \cdot 2\hat{y}$$

````python
w = 2 # Try different values
w_tensor = torch.tensor(w, dtype=float, requires_grad=True)
y_tensor = 3*w_tensor + 1    # x = 2
f = y_tensor**2

# calculate the gradient
f.backward()

print('w: ', w_tensor)
print('y: ', y_tensor)
print('f: ', f)
print('df/dw: ', w_tensor.grad)   # x*2*y_hat = 2*2*5 = 20
````

### 예제 4: 신경망 + MSE 손실 — $\ell = (y - \sigma)^2 / 2$

간단한 신경망(입력 2개 + sigmoid 활성화)을 생각합니다.

$$z = w_1 x_1 + w_2 x_2 + b, \qquad \sigma(z) = \frac{1}{1+e^{-z}}, \qquad \ell = \frac{(y - \sigma)^2}{2}$$

입력·파라미터: $x_1 = 2,\; x_2 = 1,\; w_1 = 10,\; w_2 = 10,\; b = -25,\; y = 0$.

체인 룰로 $\dfrac{d\ell}{dw_1}$을 유도하면:

$$\frac{d\ell}{dw_1} = \frac{d\ell}{d\sigma} \cdot \frac{d\sigma}{dz} \cdot \frac{dz}{dw_1} = -(y-\sigma) \cdot \sigma(1-\sigma) \cdot x_1$$

$z = 5$이므로 $\sigma \approx 0.9933$로 1에 매우 가깝고(정답 $y=0$과 정반대로 예측), $\sigma(1-\sigma) \approx 0.0067$이 매우 작기 때문에 **예측이 틀렸음에도 MSE gradient가 거의 사라집니다**(vanishing gradient 현상).

````python
x1, x2, y_true = 2.0, 1.0, 0.0

w1 = torch.tensor(10.0, requires_grad=True)
w2 = torch.tensor(10.0, requires_grad=True)
b  = torch.tensor(-25.0, requires_grad=True)

z = w1*x1 + w2*x2 + b
sigma = 1 / (1 + torch.exp(-z))
loss = (y_true - sigma)**2 / 2

loss.backward()

print('z:     ', z.item())
print('sigma: ', sigma.item())
print('loss:  ', loss.item())
print('dL/dw1:', w1.grad.item())   # -(y-sigma) * sigma*(1-sigma) * x1
````

### 예제 5: 신경망 + BCE 손실 — $\ell = -\bigl(y\log\sigma + (1-y)\log(1-\sigma)\bigr)$

같은 신경망에서 손실만 **이진 교차 엔트로피(BCE)**로 바꿉니다.

$$\ell = -\bigl(y\log\sigma + (1-y)\log(1-\sigma)\bigr)$$

체인 룰 유도:

$$\frac{d\ell}{d\sigma} = -\Bigl(\frac{y}{\sigma} - \frac{1-y}{1-\sigma}\Bigr) = \frac{\sigma - y}{\sigma(1-\sigma)}$$

$$\frac{d\ell}{dw_1} = \frac{d\ell}{d\sigma} \cdot \frac{d\sigma}{dz} \cdot \frac{dz}{dw_1} = (\sigma - y) \cdot x_1$$

BCE에서는 $\sigma(1-\sigma)$ 항이 깔끔하게 상쇄되어, **예제 4의 MSE보다 gradient가 훨씬 크게** 나타납니다(특히 $\sigma$가 0/1에 가까울 때). 같은 입력·파라미터($x_1=2, x_2=1, w_1=w_2=10, b=-25, y=0$)로 비교해 봅니다.

````python
x1, x2, y_true = 2.0, 1.0, 0.0

w1 = torch.tensor(10.0, requires_grad=True)
w2 = torch.tensor(10.0, requires_grad=True)
b  = torch.tensor(-25.0, requires_grad=True)

z = w1*x1 + w2*x2 + b
sigma = 1 / (1 + torch.exp(-z))
loss = -(y_true * torch.log(sigma) + (1 - y_true) * torch.log(1 - sigma))

loss.backward()

print('z:     ', z.item())
print('sigma: ', sigma.item())
print('loss:  ', loss.item())
print('dL/dw1:', w1.grad.item())   # (sigma - y) * x1
````

### 예제 6: 모든 $w$에 대한 gradient — 출력 라인만 추가하면 끝

`loss.backward()` 한 번이면 **그래프에 연결된 모든 `requires_grad=True` 텐서**의 gradient가 함께 계산됩니다. 따라서 $w_1$뿐 아니라 $w_2, b$의 기울기까지 얻고 싶다면 코드 구조를 바꿀 필요 없이 **출력 라인만 추가**하면 됩니다.

$$\frac{d\ell}{dw_1} = (\sigma - y)\,x_1, \qquad \frac{d\ell}{dw_2} = (\sigma - y)\,x_2, \qquad \frac{d\ell}{db} = (\sigma - y)$$

````python
x1, x2, y_true = 2.0, 1.0, 0.0

w1 = torch.tensor(10.0, requires_grad=True)
w2 = torch.tensor(10.0, requires_grad=True)
b  = torch.tensor(-25.0, requires_grad=True)

z = w1*x1 + w2*x2 + b
sigma = 1 / (1 + torch.exp(-z))
loss = -(y_true * torch.log(sigma) + (1 - y_true) * torch.log(1 - sigma))

loss.backward()

print('z:     ', z.item())
print('sigma: ', sigma.item())
print('loss:  ', loss.item())
print('dL/dw1:', w1.grad.item())   # (sigma - y) * x1
print('dL/dw2:', w2.grad.item())   # (sigma - y) * x2
print('dL/db: ', b.grad.item())    # (sigma - y)
````

### 예제 7: $\mathbf{w}$를 벡터로 구성 — 한 번에 처리

$w_1, w_2$를 각각 스칼라 텐서로 따로 두는 대신 **벡터 하나**로 묶으면, 파라미터가 아무리 많아져도 코드는 거의 변하지 않습니다. 우리가 쓰는 `z = Xw` 매트릭스 컨벤션(입력은 행, 가중치는 열)을 그대로 사용합니다.

$$\mathbf{X} = \begin{bmatrix} x_1 & x_2 \end{bmatrix}\;(1 \times 2), \qquad \mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \end{bmatrix}\;(2 \times 1), \qquad z = \mathbf{X}\mathbf{w} + b$$

`backward()` 호출 후 `w.grad`는 `(2,)` 텐서로, $w_1, w_2$ 각각의 기울기를 한 번에 담습니다.

````python
X = torch.tensor([[2.0, 1.0]])                       # (1, 2) — 1 sample × 2 features (행 벡터)
y_true = 0.0

w = torch.tensor([10.0, 10.0], requires_grad=True)   # (2,)
b = torch.tensor(-25.0, requires_grad=True)

z = X @ w + b                                         # (1,) — 한 샘플의 z. 여러샘플시 vector가 됨
sigma = 1 / (1 + torch.exp(-z))
loss = -(y_true * torch.log(sigma) + (1 - y_true) * torch.log(1 - sigma))

loss.sum().backward()   # 여러 샘플 loss를 집계하여 scalar objective를 만들고 backward

print('z:     ', z.item())
print('sigma: ', sigma.item())
print('loss:  ', loss.item())
print('dL/dw: ', w.grad)         # tensor([dL/dw1, dL/dw2]) — 한 번에 다 나옴
print('dL/db: ', b.grad.item())
````

### 예제 8: bias를 $\mathbf{X}$에 흡수 — 증강 표기

$\mathbf{X}$의 맨 앞에 `1` 컬럼을 추가하고 bias도 $\mathbf{w}$의 0번 성분으로 흡수하면 **bias 항이 사라집니다**. Section 3에서 사용하는 $\mathbf{X} \in \mathbb{R}^{N \times (d+1)}$ 컨벤션과 같은 방식입니다.

$$\mathbf{X} = \begin{bmatrix} 1 & x_1 & x_2 \end{bmatrix}\;(1 \times 3), \qquad \mathbf{w} = \begin{bmatrix} b \\ w_1 \\ w_2 \end{bmatrix}\;(3 \times 1), \qquad z = \mathbf{X}\mathbf{w}$$

`w.grad`의 0번 성분이 bias gradient, 1·2번 성분이 $w_1, w_2$의 gradient로 **한 텐서에 모두 담깁니다**. 예제 6·7과 완전히 동일한 값이 나오는지 확인합니다.

````python
X = torch.tensor([[1.0, 2.0, 1.0]])                  # (1, 3) — [1, x1, x2] (맨 앞 1은 bias용)
y_true = 0.0

w = torch.tensor([-25.0, 10.0, 10.0], requires_grad=True)   # (3,) — [b, w1, w2]

z = X @ w                                             # (1,) — bias 항 없음
sigma = 1 / (1 + torch.exp(-z))
loss = -(y_true * torch.log(sigma) + (1 - y_true) * torch.log(1 - sigma))

loss.sum().backward()

print('z:     ', z.item())
print('sigma: ', sigma.item())
print('loss:  ', loss.item())
print('dL/dw: ', w.grad)   # tensor([dL/db, dL/dw1, dL/dw2]) — bias gradient도 포함
````

# 3. Autograd로 Gradient Descent 구현

Section 2에서 `backward()`가 미분을 자동으로 계산해 주는 것을 확인했습니다.  
이제 이를 활용하여 **실제 gradient descent**를 구현합니다.

간단한 toy 데이터로 선형 회귀를 세 가지 방법으로 학습하고, **결과가 동일**함을 확인합니다.

$$\hat{y} = \mathbf{Xw}, \qquad J = \tfrac{1}{2}\|\mathbf{Xw} - \mathbf{y}\|_2^2, \qquad \mathbf{w} \leftarrow \mathbf{w} - \rho\, \partial J / \partial \mathbf{w}$$

### 데이터 준비

````python
# toy data: y = 0.3*x1 + 0.5*x2 + 0.1 + noise  (2-feature regression)
np.random.seed(42)
X_raw = np.random.randn(50, 2)                        # (50, 2) — raw feature [x1, x2]
y = X_raw @ np.array([[0.3], [0.5]]) + 0.1 + 0.05 * np.random.randn(50, 1)

X = np.hstack([np.ones((50, 1)), X_raw])              # (50, 3) — [x0=1, x1, x2] (bias 증강)

X_t     = torch.tensor(X,     dtype=torch.float32)
X_raw_t = torch.tensor(X_raw, dtype=torch.float32)
y_t     = torch.tensor(y,     dtype=torch.float32)

w_init = np.array([[0.0], [0.0], [0.0]])              # (3, 1) — [b, w1, w2]
rho = 0.02
n_iter = 200

print(f'X: {X.shape}, X_raw: {X_raw.shape}, y: {y.shape}')
````

### 모델 정의

실제 학습 코드에서는 **모델을 먼저 함수(혹은 `nn.Module` 클래스)로 정의**한 뒤, 학습 루프에서 "모델 호출 → 손실 계산 → 역전파" 형태로 사용합니다. 여기서는 가장 단순한 선형 모델을 함수로 만들어 둡니다. numpy 배열과 torch 텐서 모두에 그대로 통합니다.

$$\text{model}(\mathbf{X}, \mathbf{w}) = \mathbf{X}\mathbf{w}$$

````python
def model(X, w):
    # 선형 모델
    y_hat = X @ w
    return y_hat
````

### 방법 1: numpy — 수식으로 gradient 계산

$$J = \tfrac{1}{2}\|\mathbf{Xw} - \mathbf{y}\|_2^2 \;\Rightarrow\; \frac{\partial J}{\partial \mathbf{w}} = \mathbf{X}^T(\mathbf{Xw} - \mathbf{y}) = -\mathbf{X}^T\mathbf{E}$$

> $J$ 를 $\tfrac{1}{2}\|\mathbf{E}\|_2^2$ 로 정의하면 미분에서 상수 2가 사라져 기울기가 $-\mathbf{X}^T\mathbf{E}$ 로 깔끔해진다.
> autograd 도 같은 $J$ 를 미분하므로 방법 2의 결과와 정확히 일치한다.

````python
w_np = w_init.copy()
loss_hist_np = []

for i in range(n_iter):
    y_hat = model(X, w_np)
    loss = ((y_hat - y)**2).sum() / 2
    E    = y - y_hat
    dJ   = -X.T @ E
    w_np = w_np - rho * dJ
    loss_hist_np.append(loss)
    if i % 50 == 0:
        print(f'iter {i:3d} | loss = {loss:.4f} | w = {w_np.flatten()}')
````

### 방법 2: pytorch — autograd로 gradient 자동 계산

수식 유도 없이 `loss.backward()` 한 줄이면 됩니다.

> **`w.grad.zero_()`가 필요한 이유:**  
> PyTorch는 `backward()` 호출 시 gradient를 **누적**(accumulate)합니다.  
> 매 iteration마다 이전 gradient를 초기화하지 않으면 값이 계속 쌓입니다.

````python
w_t = torch.tensor(w_init, dtype=torch.float32, requires_grad=True)
loss_hist_torch = []

for i in range(n_iter):
    y_hat = model(X_t, w_t)
    loss = ((y_hat - y_t)**2).sum() / 2
    loss.backward()
    with torch.no_grad():
        w_t -= rho * w_t.grad
    w_t.grad.zero_()
    loss_hist_torch.append(loss.item())
    if i % 50 == 0:
        print(f'iter {i:3d} | loss = {loss.item():.4f} | w = {w_t.detach().numpy().flatten()}')
````

### 결과 비교

````python
print('w (numpy) :', w_np.flatten())
print('w (torch) :', w_t.detach().numpy().flatten())
````

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# 왼쪽: parity plot (y vs y_hat) — feature가 2개 이상일 때 일반적으로 쓰는 그림
y_hat = model(X, w_np)
axes[0].scatter(y, y_hat, c='r', s=20, alpha=0.7)
lo, hi = float(y.min()), float(y.max())
axes[0].plot([lo, hi], [lo, hi], 'k--', lw=1)
axes[0].set_title('Parity plot (y vs y_hat)')
axes[0].set_xlabel('y (true)'); axes[0].set_ylabel('y_hat')
axes[0].grid(alpha=0.3)

axes[1].plot(loss_hist_np, 'b-', lw=3, label='numpy (formula)')
axes[1].plot(loss_hist_torch, 'r--', lw=2, label='torch (autograd)')
axes[1].set_title('Loss'); axes[1].set_xlabel('iter')
axes[1].legend(); axes[1].grid(alpha=0.3)

plt.tight_layout(); plt.show()
````

# 4. torch.optim 활용

방법 2에서 `w -= rho * w.grad`로 가중치를 직접 갱신했습니다.  
`torch.optim`을 사용하면 이 갱신 과정마저도 `optimizer.step()` 한 줄로 대체됩니다.

> 지금은 단순한 SGD지만, 나중에 Adam, RMSprop 등 다양한 optimizer를 사용할 때  
> 학습 코드를 바꿀 필요 없이 **optimizer 선언만 교체**하면 됩니다.

````python
w_optim = torch.tensor(w_init, dtype=torch.float32, requires_grad=True)
optimizer = torch.optim.SGD([w_optim], lr=rho)
loss_hist_optim = []

for i in range(n_iter):
    y_hat = model(X_t, w_optim)
    loss = ((y_hat - y_t)**2).sum() / 2
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    loss_hist_optim.append(loss.item())
    if i % 50 == 0:
        print(f'iter {i:3d} | loss = {loss.item():.4f} | w = {w_optim.detach().numpy().flatten()}')
````

````python
plt.figure(figsize=(7, 4))
plt.plot(loss_hist_np, 'b-', lw=3, label='numpy (formula)')
plt.plot(loss_hist_torch, 'r--', lw=2, label='torch (manual update)')
plt.plot(loss_hist_optim, 'g:', lw=2, label='torch (optimizer.step)')
plt.title('Loss Comparison (3 methods)')
plt.xlabel('iter'); plt.ylabel('loss')
plt.legend(); plt.grid(alpha=0.3)
plt.tight_layout(); plt.show()
````

# 5. torch.nn 손실 함수 활용

Section 4에서 가중치 갱신이 `optimizer.step()` 한 줄이 됐습니다. 이번에는 **손실 정의**까지 `torch.nn.MSELoss`로 대체합니다.

이제 학습 루프에 남는 것은:
- forward: `X_t @ w`
- loss: `mse(pred, target) / 2` ← 제곱·합산은 라이브러리가 계산 (½ 은 $J$ 정의)
- backward + step: `loss.backward(); optimizer.step(); optimizer.zero_grad()`

> MSE 외에도 `nn.BCELoss`, `nn.CrossEntropyLoss` 등 **손실별 대응 클래스**를 바꿔 끼우면 수식 유도 없이 곧바로 사용할 수 있습니다.

````python
w_optim = torch.tensor(w_init, dtype=torch.float32, requires_grad=True)
optimizer = torch.optim.SGD([w_optim], lr=rho)
mse_loss = torch.nn.MSELoss(reduction='sum')   # ‖E‖² 합산
loss_hist_nn = []

for i in range(n_iter):
    y_hat = model(X_t, w_optim)
    loss = mse_loss(y_hat, y_t) / 2   # J = ½‖E‖² (앞 방법들과 동일한 정의)

    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    loss_hist_nn.append(loss.item())

    if i % 50 == 0:
        print(f'iter {i:3d} | loss = {loss.item():.4f} | w = {w_optim.detach().numpy().flatten()}')

print(f'\nw (nn.MSELoss) : {w_optim.detach().numpy().flatten()}')
````

### 자주 쓰는 PyTorch 손실 함수

| 용도 | 클래스 | 출력 | 비고 |
|---|---|---|---|
| 회귀 | `nn.MSELoss` | $\hat{y}$ | $(\hat{y}-y)^2$ 평균/합 |
| 회귀 | `nn.L1Loss` | $\hat{y}$ | $\lvert \hat{y}-y \rvert$ 평균, outlier에 덜 민감 |
| 이진 분류 | `nn.BCELoss` | probability (sigmoid 후) | — |
| 이진 분류 | `nn.BCEWithLogitsLoss` | $z$ (sigmoid 전단계) | BCELoss + sigmoid |
| 다중 분류 | `nn.NLLLoss` | log-probability (log_softmax 후) | — |
| 다중 분류 | `nn.CrossEntropyLoss` | $z$ (softmax 전단계) | NLLLoss + log_softmax |

**패턴.** `WithLogits` / `CrossEntropy` 처럼 **활성 함수(sigmoid · log_softmax)가 손실에 내장된 버전**은 모델의 raw $z$를 그대로 받습니다. 반대로 기본 버전(`BCELoss`, `NLLLoss`)은 **활성 함수가 이미 적용된 값**을 모델이 출력해서 넘겨줘야 합니다. 내장된 버전이 수치적으로 더 안정적이므로 실제 학습 코드에서는 이쪽을 더 많이 씁니다.

> 모든 손실 클래스는 `reduction='mean' | 'sum' | 'none'` 인자로 축소 방식을 선택할 수 있습니다 (기본값은 `'mean'`).

### 학습 루프를 함수로 — 재사용 준비 완료

이제 **모델 호출 · 손실 · backward · optimizer.step**이 모두 준비되었으니, `for` 루프 자체를 함수로 묶어두면 다음 섹션부터는 **모델만 바꿔 가며 한 줄로 학습**할 수 있습니다.

````python
def train(model, X, y, optimizer, loss_fn, n_iter):
    # 모델을 n_iter번 학습하고 iteration별 loss 리스트를 반환
    loss_hist = []
    for i in range(n_iter):
        y_hat = model(X)
        loss = loss_fn(y_hat, y)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
        loss_hist.append(loss.item())
        if i % 50 == 0:
            print(f'iter {i:3d} | loss = {loss.item():.4f}')
    return loss_hist


# 손실 함수: J = ½‖E‖²  (앞의 방법들과 동일한 정의)
mse_sum = torch.nn.MSELoss(reduction='sum')
def half_mse(pred, target):
    return mse_sum(pred, target) / 2
````

---
# 6. 모델 구조도 PyTorch로 — nn.Linear

지금까지는 `w = torch.tensor(..., requires_grad=True)`로 파라미터 텐서를 직접 만들고 `y_hat = X @ w`로 forward를 계산했습니다. PyTorch에는 이 선형 변환을 캡슐화한 **`nn.Linear`** 모듈이 있습니다.

```
model = nn.Linear(in_features, out_features)
```

- `nn.Linear(1, 1)` → 입력 1차원, 출력 1차원
- `nn.Linear(2, 1)` → 입력 2차원 ($x_1, x_2$), 출력 1차원 — **지금 데이터**
- `nn.Linear(4, 3)` → 입력 4차원, 출력 3차원 (다변량 회귀·다중 분류)

모델 구조의 **숫자만 바꾸면** 나머지 학습 코드는 그대로 쓸 수 있습니다. `model.weight`, `model.bias`가 자동 생성되고, `model.parameters()`로 optimizer에 한 번에 넘겨집니다. 이제 입력은 bias 컬럼 없는 raw `X_raw_t` (50, 2)를 사용합니다.

> 기본 초기화는 Kaiming uniform이라 시작점이 이전 섹션들(0으로 시작)과 다릅니다. 이전과 동일한 loss 곡선을 보기 위해 여기서는 파라미터를 명시적으로 0으로 초기화합니다.

````python
model = torch.nn.Linear(2, 1)                # 입력 2차원 → 출력 1차원 (bias 포함)
with torch.no_grad():                        # 기본 random init 대신 0으로 시작 (이전 섹션과 비교용)
    model.weight.zero_()
    model.bias.zero_()
````

````python
optimizer = torch.optim.SGD(model.parameters(), lr=rho)
loss_fn = half_mse
loss_hist_linear = train(model, X_raw_t, y_t, optimizer, loss_fn, n_iter)

print(f'\nfinal: w = {model.weight.detach().numpy().flatten()}, b = {model.bias.item():.4f}')
````

# 7. 모델을 클래스로 정의 — nn.Module

Section 6에서는 `nn.Linear`를 직접 사용했지만, 모델이 복잡해지면 **클래스로 정의**하는 것이 표준 패턴입니다.

```python
class ModelName(nn.Module):              # nn.Module을 상속
    def __init__(self, ...):              # 레이어(학습 파라미터) 정의
        super().__init__()                # nn.Module 초기화 (필수)
        self.layer = nn.Linear(...)

    def forward(self, x):                 # 순전파 계산 정의
        return self.layer(x)
```

| 구성요소 | 역할 |
|---|---|
| `nn.Module` 상속 | 파라미터 추적, gradient 계산 등 PyTorch 기능 활용 |
| `__init__` | `nn.Linear` 등 학습 가능한 레이어를 `self` 속성으로 등록 |
| `super().__init__()` | `nn.Module` 초기화 — **없으면 파라미터 추적 불가** |
| `forward(x)` | 입력 `x`에서 출력까지의 계산 과정을 정의 |
| `model(x)` | 내부적으로 `model.forward(x)` 실행 (직접 `.forward()` 호출하지 않음) |

> 이 패턴은 `self.fc1 = nn.Linear(...)`, `self.fc2 = nn.Linear(...)` 처럼 층만 더 붙이면 본격적인 MLP로 그대로 확장됩니다.

````python
class LinearModel(torch.nn.Module):
    def __init__(self, in_features, out_features):
        super().__init__()
        self.linear = torch.nn.Linear(in_features, out_features)

    def forward(self, x):
        y_hat = self.linear(x)
        return y_hat

model = LinearModel(2, 1)
with torch.no_grad():
    model.linear.weight.zero_()
    model.linear.bias.zero_()
````

````python
optimizer = torch.optim.SGD(model.parameters(), lr=rho)
loss_fn = half_mse
loss_hist_class = train(model, X_raw_t, y_t, optimizer, loss_fn, n_iter)

print(f'\nfinal: w = {model.linear.weight.detach().numpy().flatten()}, b = {model.linear.bias.item():.4f}')
````

---
## 다음 장 맛보기 — nn.Linear 여러 개 붙이면 MLP

`nn.Module` 클래스 구조를 그대로 두고 **`__init__`에 Linear 층을 더 쌓고 사이에 활성 함수를 끼워 넣기만 하면** 본격적인 **MLP (Multi-Layer Perceptron)**가 됩니다. 층을 아무리 깊게 쌓아도 학습 코드(`train(...)`)는 **하나도 바꾸지 않고** 그대로 재사용됩니다.

(층이 $L$개일 때) $\mathbf{h}_0 = \mathbf{X}, \qquad \mathbf{h}_\ell = \text{ReLU}(\mathbf{h}_{\ell-1}\mathbf{W}_\ell + \mathbf{b}_\ell) \quad (\ell = 1, \dots, L{-}1), \qquad \hat{\mathbf{y}} = \mathbf{h}_{L-1}\mathbf{W}_L + \mathbf{b}_L$

````python
class MLP(torch.nn.Module):
    def __init__(self, in_features, hidden, out_features):
        super().__init__()
        self.fc1  = torch.nn.Linear(in_features, hidden)
        self.fc2  = torch.nn.Linear(hidden, hidden)
        self.fc3  = torch.nn.Linear(hidden, hidden)
        self.fc4  = torch.nn.Linear(hidden, out_features)
        self.relu = torch.nn.ReLU()

    def forward(self, x):
        h1    = self.relu(self.fc1(x))
        h2    = self.relu(self.fc2(h1))
        h3    = self.relu(self.fc3(h2))
        y_hat = self.fc4(h3)
        return y_hat

torch.manual_seed(42)
mlp = MLP(2, 32, 1)              # 입력 2 → 은닉 32 → 32 → 32 → 출력 1
print(mlp)
print(f'파라미터 수: {sum(p.numel() for p in mlp.parameters())}')

# 학습 코드는 그대로. optimizer만 Adam으로 교체 (깊은 네트워크는 SGD보다 Adam이 안정적)
optimizer = torch.optim.Adam(mlp.parameters(), lr=0.01)
loss_fn = half_mse
loss_hist_mlp = train(mlp, X_raw_t, y_t, optimizer, loss_fn, n_iter)

print(f'\nMLP    final loss : {loss_hist_mlp[-1]:.4f}')
print(f'Linear final loss : {loss_hist_class[-1]:.4f}')
````
